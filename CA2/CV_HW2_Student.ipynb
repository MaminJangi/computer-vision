## ⁦A1⁩ — شناخت داده و بازتولید اختلال‌ها — ⁦8⁩ نمره

تمام ورودی‌ها را بخوانید و جدولی از شناسه و ابعاد و نوع داده و دامنهٔ شدت و مجموعهٔ ⁦dev/eval⁩ بسازید
برای سه نمونهٔ متفاوت تصویر قبل و بعد از اختلال را کنار هم نمایش دهید و برای هر نمونه مشخص کنید متن روشن است یا تیره و کدام بخش ظریف متن در پیش‌پردازش ممکن است از بین برود
با خواندن تنظیمات اختلال یک نمونه را با همان بذر بازتولید و برابری پیکسل‌به‌پیکسل آن را بررسی کنید
تفاوت چگالی نقاط انتخاب‌شده برای نویز و درصد پیکسل‌هایی که واقعاً تغییر کرده‌اند را توضیح دهید

**تحویل این قسمت:** جدول ۲۰ ورودی و سه مقایسهٔ تصویری و نتیجهٔ بازتولید  
بارم: شناخت داده ۳ — بازتولید ۳ — توضیح ۲

</div>

```python
# A1 — Read image sizes, types, and intensity ranges
A1_OUTPUT = OUTPUT / "A1"
A1_OUTPUT.mkdir(exist_ok=True)

inventory = []
for sample in samples:
    with Image.open(DATA / sample["input_path"]) as image:
        pixels = np.asarray(image)
    height, width = pixels.shape
    inventory.append({
        "sample_id": sample["sample_id"],
        "width": width,
        "height": height,
        "dtype": str(pixels.dtype),
        "min_intensity": int(pixels.min()),
        "max_intensity": int(pixels.max()),
        "split": sample["split"],
    })
```

```python
# A1 — Save and display the inventory table
with (A1_OUTPUT / "inventory.csv").open("w", encoding="utf-8", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=list(inventory[0]))
    writer.writeheader()
    writer.writerows(inventory)

headers = ("Sample", "Width (px)", "Height (px)", "dtype", "Min", "Max", "Split")
table_head = "".join(f"<th>{header}</th>" for header in headers)
table_rows = "".join(
    "<tr>" + "".join(f"<td>{value}</td>" for value in row.values()) + "</tr>"
    for row in inventory
)
display(HTML(f'<table dir="ltr"><thead><tr>{table_head}</tr></thead><tbody>{table_rows}</tbody></table>'))
```

<div dir="rtl" align="right" style="direction: rtl; text-align: right;">
<p><strong>خواندن جدول.</strong> یک تصویر خاکستری را می‌توان مانند جدول اعداد دید. هر عدد شدت روشنایی یک پیکسل است. شکل آرایه ابتدا ارتفاع و سپس عرض را می‌دهد. نوع <bdi dir="ltr">uint8</bdi> یعنی هر پیکسل یک عدد صحیح بین ۰ و ۲۵۵ دارد. هر ۲۰ ورودی رسمی هر دو حد را دارند. وجود این دو مقدار به‌تنهایی دربارهٔ قطبیت متن چیزی نمی‌گوید، چون خود نویز هم می‌تواند پیکسل سیاه یا سفید بسازد.</p>
</div>

```python
# A1 — Show original and corrupted images side by side
example_ids = ("S001", "S002", "S012")
dev_by_id = {row["sample_id"]: row for row in samples if row["split"] == "dev"}
fig, axes = plt.subplots(3, 2, figsize=(8.0, 3.2), layout="constrained")

for sample_id, row_axes in zip(example_ids, axes):
    sample = dev_by_id[sample_id]
    for ax, field in zip(row_axes, ("original_path", "input_path")):
        with Image.open(DATA / sample[field]) as image:
            gray = np.asarray(image.convert("L"))
        ax.imshow(gray, cmap="gray", vmin=0, vmax=255, interpolation="nearest")
        ax.set_axis_off()
    row_axes[0].annotate(sample_id, xy=(0, 0.5), xycoords="axes fraction",
                         xytext=(-9, 0), textcoords="offset points",
                         ha="right", va="center", fontsize=9, color="#202833")

axes[0, 0].set_title("Before synthetic corruption", fontsize=11)
axes[0, 1].set_title("Official input", fontsize=11)
fig.savefig(A1_OUTPUT / "before_after.png", dpi=220, facecolor="white")
plt.show()
plt.close(fig)
```

```python
# A1 — Zoom in on thin strokes
# Locate the specific strokes discussed below.
detail_fig, detail_axes = plt.subplots(1, 2, figsize=(8, 2.6), layout="constrained")
details = (
    ("S001", "original_path", "S001 | diagonal strokes of the two 4s",
     (13, 40), (26, 3), ((19, 12), (32, 12))),
    ("S002", "input_path", "S002 | middle stroke of the first 3",
     (17, 34), (17, 0), ((26, 7),)),
)

for ax, (sample_id, field, title, xlim, ylim, points) in zip(detail_axes, details):
    with Image.open(DATA / dev_by_id[sample_id][field]) as image:
        gray = np.asarray(image.convert("L"))
    ax.imshow(gray, cmap="gray", vmin=0, vmax=255, interpolation="nearest")
    ax.set(xlim=xlim, ylim=ylim, title=title)
    ax.set_axis_off()
    for x, y in points:
        ax.annotate("", xy=(x, y), xytext=(x - 4, y - 4),
                    arrowprops={"arrowstyle": "->", "color": "#1a9ca6", "lw": 2})

detail_fig.savefig(A1_OUTPUT / "character_details.png", dpi=200, facecolor="white")
plt.show()
plt.close(detail_fig)
```

## ⁦A2⁩ — مقایسهٔ دو مسیر پیش‌پردازش — ⁦12⁩ نمره

دو مسیر متمایز برای آماده‌سازی تصویر طراحی کنید و روی ورودی‌های یکسان مقایسه کنید
برای نمونه می‌توانید مسیر مبتنی بر ⁦Median⁩ را با مسیر مبتنی بر ⁦Gaussian⁩ و بهبود کنتراست مقایسه کنید اما انتخاب با شماست
کد مناسب تمرین اول خود را می‌توانید بازاستفاده کنید
اگر تغییر کنتراست یا تیزسازی را لازم دانستید اثرش را نشان دهید و توضیح دهید چرا آن را اضافه کرده‌اید
تیزسازی را معادل بازگرداندن اطلاعات ازدست‌رفته بر اثر تاری در نظر نگیرید

روش مدیریت مرز و پارامترها را ثبت کنید و انتخاب را فقط روی ⁦dev⁩ انجام دهید
برای سه نمونه شدت‌ها یا هیستوگرام‌ها و شکل حروف و نشانه‌ها را قبل و بعد مقایسه کنید
فقط کم‌شدن نویز ظاهری دلیل کافی برای انتخاب مسیر نیست

**تحویل این قسمت:** پیاده‌سازی هر دو مسیر و تصاویر میانی و جدول پارامترها و دلیل انتخاب  
بارم: مسیر اول ۴ — مسیر دوم ۴ — مقایسه و استدلال ۴

</div>

```python
# A2.1 — Two preprocessing routes
A2_ROUTES = {
    "A": {"filter": "median", "size": 3, "mode": "reflect"},
    "B": {"filter": "gaussian", "sigma": 0.65, "radius": 2,
          "mode": "reflect", "stretch_percentiles": (2, 98)},
}


def preprocess_image(gray, config):
    """Return the preprocessed grayscale image and any intermediate results."""
    pixels = np.asarray(gray)
    if pixels.ndim != 2 or pixels.dtype != np.uint8 or pixels.size == 0:
        raise ValueError("Expected a non-empty, 8-bit single-channel image.")

    if config["filter"] == "median":
        filtered = ndimage.median_filter(pixels, size=config["size"], mode=config["mode"])
        return filtered, {}

    if config["filter"] != "gaussian":
        raise ValueError(f"Unknown preprocessing filter: {config['filter']}")

    smoothed = ndimage.gaussian_filter(
        pixels.astype(np.float32), sigma=config["sigma"],
        radius=config["radius"], mode=config["mode"],
    )
    low, high = np.percentile(smoothed, config["stretch_percentiles"])
    stretched = smoothed
    if high > low:
        stretched = np.clip((smoothed - low) * (255 / (high - low)), 0, 255)
    return np.rint(stretched).astype(np.uint8), {
        "smoothed": smoothed,
        "stretch_limits": (float(low), float(high)),
    }
```

<div dir="rtl" align="right" style="direction: rtl; text-align: right;">
<p><strong>مسیر A.</strong> فیلتر <bdi dir="ltr">Median</bdi> نه شدت در پنجره‌ی <bdi dir="ltr">3 × 3</bdi> را مرتب می‌کند و عدد میانی را جای پیکسل مرکزی می‌گذارد. یک نقطه‌ی کاملاً سیاه یا سفید وقتی همسایه‌های معمولی دارد، معمولاً از نتیجه حذف می‌شود. پنجره‌ی کوچک را برای کاهش خطر حذف بخش‌های باریک انتخاب کردیم.</p>
<p><strong>مسیر B.</strong> فیلتر <bdi dir="ltr">Gaussian</bdi> از میانگین وزن‌دار استفاده می‌کند. پیکسل‌های نزدیک‌تر وزن بیشتری دارند. انحراف معیار <bdi dir="ltr">σ = 0.65</bdi> و شعاع ۲ را انتخاب کردیم تا تغییر روی اطراف هر پیکسل محدود باشد. سپس در هر تصویر صدک‌های ۲ و ۹۸ را به صفر و ۲۵۵ نگاشتیم. شدت‌های بین این دو حد به‌تدریج پخش می‌شوند و شدت‌های بیرون آن‌ها برش می‌خورند. اگر دو حد برابر باشند، کشش کنتراست را رد می‌کنیم.</p>
<p>در هر دو مسیر، گزینه‌ی <bdi dir="ltr">reflect</bdi> مقادیر لبه را برای محاسبه‌ی همسایگی بازتاب می‌دهد. خروجی هر مسیر یک تصویر خاکستری با اندازه‌ی ورودی است. در این بخش تیزسازی انجام نشده است.</p>
</div>

<div dir="rtl" align="right" style="direction: rtl; text-align: right;">
<p><strong>تنظیم روشنایی در A1.</strong> فرمول <bdi dir="ltr"><code>clip(0.82 × I + 12, 0, 255)</code></bdi> روی هر پیکسل به‌تنهایی اجرا می‌شود. برای نمونه، اگر <bdi dir="ltr">I = 100</bdi> باشد، مقدار جدید ۹۴ است. با همین فرمول صفر به ۱۲ و ۲۵۵ پس از تبدیل به عدد صحیح به ۲۲۱ می‌رسد.</p>
<p><strong>محاسبه‌ی Gaussian.</strong> ابتدا وزن فاصله‌ی <bdi dir="ltr"><code>k</code></bdi> در یک جهت را از <bdi dir="ltr"><code>exp(−k² / (2σ²))</code></bdi> می‌گیریم و وزن‌ها را طوری مقیاس می‌کنیم که جمعشان یک شود. در مسیر <bdi dir="ltr">B</bdi> وزن مرکزی در یک جهت حدود ۰٫۶۱۳۵ است. فیلتر در هر دو جهت اعمال می‌شود و وزن پیکسل مرکزی حدود ۰٫۳۷۶ می‌شود. اگر مرکز صفر و بقیه‌ی پنجره ۲۰۰ باشد، حاصل <bdi dir="ltr"><code>0 × 0.376 + 200 × 0.624 ≈ 125</code></bdi> است.</p>
<p><strong>صدک و کشش کنتراست.</strong> شدت‌ها را از کم به زیاد مرتب کنید. در یک تصویر با ۱۰۰ پیکسل، صدک دوم نزدیک پایین فهرست و صدک ۹۸ نزدیک بالای آن است. این درصدها به <em>تعداد پیکسل‌ها</em> مربوط‌اند. در <bdi dir="ltr">S010</bdi> پس از فیلتر، این دو حد تقریباً ۷۷٫۷۵ و ۱۷۱٫۱۸ بودند. فرمول <bdi dir="ltr"><code>clip(255 × (G − P2) / (P98 − P2), 0, 255)</code></bdi> شدت ۱۲۵ را تقریباً به ۱۲۹ می‌رساند. دو صدک باعث می‌شوند چند پیکسل خیلی تیره یا روشن به‌تنهایی محدوده‌ی کشش را تعیین نکنند.</p>

```python
# A3 — Configure the mask rules and define Otsu thresholding
A3_CONFIG = {
    "border_fraction": 0.06,
    "line_min_width_fraction": 0.80,
    "line_outer_rows_fraction": 0.25,
    "min_component_area_fraction": 0.003,
    "text_band_margin_fraction": 0.10,
}


def otsu_threshold(gray):
    """Split the histogram where the two groups separate most strongly."""
    probability = np.bincount(gray.ravel(), minlength=256) / gray.size
    weight = np.cumsum(probability)
    partial_mean = np.cumsum(probability * np.arange(256))
    separation = np.zeros(256)
    np.divide(
        (partial_mean[-1] * weight - partial_mean) ** 2,
        weight * (1 - weight),
        out=separation,
        where=(weight > 0) & (weight < 1),
    )
    return int(np.argmax(separation))
```

```python
# A3 — Decide which brightness group is text
def choose_text_polarity(gray, threshold, config):
    """Identify text by comparing the dark and light groups at the border."""
    height, width = gray.shape
    dark_candidate = gray <= threshold

    # The outer ring usually contains mostly background.
    border_size = max(1, round(config["border_fraction"] * min(height, width)))
    border = np.ones(gray.shape, dtype=bool)
    border[border_size:-border_size, border_size:-border_size] = False
    dark_border_fraction = float(dark_candidate[border].mean())
    if gray.min() == gray.max():
        polarity = "undetermined"
        initial = np.zeros(gray.shape, dtype=bool)
    else:
        if dark_border_fraction == 0.5:
            dark_text = float(dark_candidate.mean()) <= 0.5
        else:
            dark_text = dark_border_fraction < 0.5
        polarity = "dark" if dark_text else "light"
        initial = dark_candidate if dark_text else ~dark_candidate
    return polarity, initial, dark_border_fraction
```

```python
# A3 — Clean long horizontal runs near the edges
def remove_horizontal_lines(initial, config):
    """Remove only long foreground runs in the outer rows."""
    height, width = initial.shape
    # Remove long continuous runs only in the top and bottom bands.
    without_lines = initial.copy()
    line_min_width = int(np.ceil(config["line_min_width_fraction"] * width))
    outer_rows = int(np.ceil(config["line_outer_rows_fraction"] * height))
    edge_rows = np.union1d(np.arange(outer_rows), np.arange(height - outer_rows, height))
    for y in edge_rows:
        edges = np.flatnonzero(np.diff(np.pad(initial[y].astype(np.int8), 1)))
        for start, stop in zip(edges[::2], edges[1::2]):
            if stop - start >= line_min_width:
                without_lines[y, start:stop] = False
    return without_lines
```

```python
# A3 — Remove small components outside the text band
def remove_outside_specks(without_lines, config):
    """Keep small components near the text band and remove distant ones."""
    height = without_lines.shape[0]
    # Keep small components in or close to the band containing larger text.
    labels, _ = ndimage.label(without_lines, structure=np.ones((3, 3), dtype=bool))
    boxes = ndimage.find_objects(labels)
    areas = np.bincount(labels.ravel())
    min_area = max(2, int(np.ceil(config["min_component_area_fraction"] * without_lines.size)))
    main_boxes = [box for label, box in enumerate(boxes, 1) if areas[label] >= min_area]
    remove = np.zeros(len(areas), dtype=bool)
    if main_boxes:
        margin = max(1, round(config["text_band_margin_fraction"] * height))
        top = min(box[0].start for box in main_boxes) - margin
        bottom = max(box[0].stop for box in main_boxes) + margin
        for label, box in enumerate(boxes, 1):
            outside_band = box[0].stop <= top or box[0].start >= bottom
            remove[label] = areas[label] < min_area and outside_band

    final = without_lines & ~remove[labels]
    return final, min_area, int(remove.sum())
```

```python
# A3 — Combine thresholding, polarity, and cleanup into one mask
def make_text_mask(filtered_gray, config, *, return_steps=False):
    """Return a same-size uint8 mask with text=255 and background=0."""
    gray = np.asarray(filtered_gray)
    if gray.ndim != 2 or gray.dtype != np.uint8 or gray.size == 0:
        raise ValueError("Expected a non-empty 8-bit grayscale image.")

    threshold = otsu_threshold(gray)
    polarity, initial, dark_border_fraction = choose_text_polarity(gray, threshold, config)
    without_lines = remove_horizontal_lines(initial, config)
    final, min_area, removed_count = remove_outside_specks(without_lines, config)
    mask = final.astype(np.uint8) * 255
    if not return_steps:
        return mask
    return mask, {
        "threshold": threshold,
        "polarity": polarity,
        "dark_border_fraction": dark_border_fraction,
        "initial": initial,
        "without_lines": without_lines,
        "line_pixels_removed": int(initial.sum() - without_lines.sum()),
        "min_area": min_area,
        "components_removed": removed_count,
        "speck_pixels_removed": int(without_lines.sum() - final.sum()),
    }
```

```python
# B1 — Find components, remove specks, and estimate text height
from PIL import ImageDraw
from scipy.optimize import linear_sum_assignment

B1_CONFIG = {
    "min_area": 2,
    "small_area_fraction": 0.04,
    "body_min_height_fraction": 0.38,
    "body_width_fallback": 0.70,
    "fragment_max_gap_fraction": 0.25,
    "fragment_min_x_overlap": 0.50,
    "fragment_max_height_fraction": 0.78,
    "wide_height_ratio": 1.30,
    "wide_typical_width_ratio": 1.45,
    "split_search_from_width": 0.60,
    "split_search_to_width": 1.48,
    "split_max_valley": 0.42,
    "split_width_penalty": 0.18,
}



def collect_character_components(foreground, config):
    """Label connected pixels and keep components near the text band."""
    labels, _ = ndimage.label(foreground, structure=np.ones((3, 3), dtype=bool))
    areas = np.bincount(labels.ravel())
    components = [
        ((x.start, y.start, x.stop, y.stop), (label,))
        for label, (y, x) in enumerate(ndimage.find_objects(labels), start=1)
        if areas[label] >= config["min_area"]
    ]
    if not components:
        return labels, [], 0.0

    bodies = [
        box for box, labels_here in components
        if box[3] - box[1] >= config["body_min_height_fraction"] * foreground.shape[0]
        and areas[labels_here[0]] >= 4
    ]
    body_height = float(np.median([b[3] - b[1] for b in (bodies or [box for box, _ in components])]))
    if bodies:
        top = min(b[1] for b in bodies) - max(1, round(0.12 * body_height))
        bottom = max(b[3] for b in bodies) + max(1, round(0.12 * body_height))
        small_area = max(config["min_area"], int(np.ceil(config["small_area_fraction"] * body_height**2)))
        components = [
            (b, ids) for b, ids in components
            if areas[ids[0]] >= small_area or (b[3] > top and b[1] < bottom)
        ]
    return labels, components, body_height
```

```python
# B1 — Merge vertically separated fragments
def merge_fragments(groups, body_height, config):
    """Join vertically separated pieces with strongly overlapping x ranges."""
    max_gap = max(1, round(config["fragment_max_gap_fraction"] * body_height))
    for i, (first, ids) in enumerate(groups):
        j = i + 1
        while j < len(groups):
            second, other_ids = groups[j]
            x1, y1, x2, y2 = first
            u1, v1, u2, v2 = second
            gap = max(y1, v1) - min(y2, v2)
            x_overlap = min(x2, u2) - max(x1, u1)
            combined = (min(x1, u1), min(y1, v1), max(x2, u2), max(y2, v2))
            if (
                0 <= gap <= max_gap
                and x_overlap >= config["fragment_min_x_overlap"] * min(x2 - x1, u2 - u1)
                and min(y2 - y1, v2 - v1) <= config["fragment_max_height_fraction"] * body_height
                and combined[3] - combined[1] <= 1.25 * body_height
            ):
                first, ids = combined, ids + other_ids
                groups[i] = (first, ids)
                groups.pop(j)
                j = i + 1
            else:
                j += 1
    return groups
```

```python
# B1 — Estimate character width and find cuts through touching shapes
def typical_character_width(groups, body_height, config):
    """Estimate nominal width from plausible single glyphs."""
    widths = [
        x2 - x1 for (x1, y1, x2, y2), _ in groups
        if y2 - y1 >= 0.65 * body_height and 0.45 * body_height <= x2 - x1 <= 1.22 * body_height
    ]
    return float(np.median(widths)) if widths else max(2, config["body_width_fallback"] * body_height)


def projection_spans(region, body_height, typical_width, config):
    """Split wide blobs at sparse columns; yield the remaining column spans."""
    projection = region.sum(axis=0)
    start, stop = 0, len(projection)
    wide_limit = max(
        body_height * config["wide_height_ratio"],
        typical_width * config["wide_typical_width_ratio"],
    )
    min_width = max(2, int(np.ceil(config["split_search_from_width"] * typical_width)))
    max_width = max(min_width, int(np.floor(config["split_search_to_width"] * typical_width)))
    while stop - start > wide_limit:
        cuts = np.arange(start + min_width, min(start + max_width, stop - min_width) + 1)
        valleys = np.minimum(projection[cuts - 1], projection[cuts]) / body_height
        eligible = valleys <= config["split_max_valley"]
        if not eligible.any():
            break
        cuts, valleys = cuts[eligible], valleys[eligible]
        scores = valleys + config["split_width_penalty"] * abs(cuts - start - typical_width) / typical_width
        cut = int(cuts[np.argmin(scores)])
        # If the valley lies just right of the cut, retain it in the left crop.
        if projection[cut] < projection[cut - 1] and cut + 1 < stop:
            cut += 1
        yield start, cut
        start = cut
    yield start, stop
```

```python
# B1 — Assemble and sort the final character boxes
def segment_characters(mask, config):
    """Return left-to-right (x,y,width,height) boxes and a same-size uint8 mask."""
    foreground = np.asarray(mask) > 0
    if foreground.ndim != 2 or not foreground.size:
        raise ValueError("Expected a non-empty single-channel text mask.")
    labels, components, body_height = collect_character_components(foreground, config)
    cleaned = np.zeros(foreground.shape, dtype=np.uint8)
    if not components:
        return [], cleaned

    groups = merge_fragments(components, body_height, config)
    typical_width = typical_character_width(groups, body_height, config)

    boxes = []
    for (x1, y1, x2, y2), ids in groups:
        region = np.isin(labels[y1:y2, x1:x2], ids)
        cleaned[y1:y2, x1:x2][region] = 255
        # A short horizontal mark, such as a hyphen, is not a wide glyph.
        spans = (projection_spans(region, body_height, typical_width, config)
                 if y2 - y1 >= 0.65 * body_height else ((0, x2 - x1),))
        for left, right in spans:
            ys, xs = np.nonzero(region[:, left:right])
            if len(xs):
                boxes.append((x1 + left + int(xs.min()), y1 + int(ys.min()),
                              int(xs.max() - xs.min() + 1), int(ys.max() - ys.min() + 1)))
    return sorted(boxes), cleaned
```

```python
# B1 — Match predicted and reference boxes by IoU
def match_character_boxes(predicted, reference, threshold=0.5):
    """Maximize the number of valid one-to-one matches, then the total IoU."""
    if not predicted or not reference:
        return []
    p = np.asarray(predicted)
    r = np.asarray(reference)
    start = np.maximum(p[:, None, :2], r[None, :, :2])
    stop = np.minimum((p[:, :2] + p[:, 2:])[:, None, :], (r[:, :2] + r[:, 2:])[None, :, :])
    intersection = np.maximum(0, stop - start).prod(axis=2)
    union = p[:, 2:].prod(axis=1)[:, None] + r[:, 2:].prod(axis=1)[None, :] - intersection
    ious = intersection / union
    eligible = ious >= threshold
    bonus = 1 + min(len(predicted), len(reference))
    scores = np.where(eligible, bonus + ious, 0)
    rows, cols = linear_sum_assignment(scores, maximize=True)
    return [(int(i), int(j), float(ious[i, j])) for i, j in zip(rows, cols) if eligible[i, j]]
```

```python
# B1 — Define helpers for the segmentation examples
from matplotlib.patches import Rectangle


def draw_mask_boxes(ax, image, boxes, color="#17a8a8", style="-"):
    ax.imshow(image, cmap="gray", vmin=0, vmax=255, interpolation="nearest")
    ax.set_axis_off()
    for x, y, w, h in boxes:
        ax.add_patch(Rectangle((x - 0.5, y - 0.5), w, h, fill=False,
                               edgecolor=color, linestyle=style, linewidth=1.6))


def component_boxes(mask):
    labels, _ = ndimage.label(mask > 0, structure=np.ones((3, 3), dtype=bool))
    return [(x.start, y.start, x.stop - x.start, y.stop - y.start)
            for y, x in ndimage.find_objects(labels)]
```

```python
# B2 — Configure the canvas and find the tight foreground bounds
B2_CONFIG = {"canvas": 32, "margin": 3}


def foreground_bounds(mask):
    """Return exclusive (left, top, right, bottom) bounds, or None if empty."""
    ys, xs = np.nonzero(mask)
    if not ys.size:
        return None
    return (int(xs.min()), int(ys.min()), int(xs.max()) + 1, int(ys.max()) + 1)
```

```python
# B2 — Resize and center a character with one shared function
def normalize_character(mask, canvas=32, margin=3):
    """Tight crop, aspect-ratio resize, and center a character on black."""
    image = np.asarray(mask)
    if image.ndim != 2:
        raise ValueError("Expected a single-channel character mask.")
    if margin < 0 or canvas <= 2 * margin:
        raise ValueError("Canvas must have room for content inside its margin.")

    result = np.zeros((canvas, canvas), dtype=np.uint8)
    foreground = image > 0
    bounds = foreground_bounds(foreground)
    if bounds is None:
        return result

    left, top, right, bottom = bounds
    tight = foreground[top:bottom, left:right].astype(np.float32) * 255
    height, width = tight.shape
    usable = canvas - 2 * margin
    scale = usable / max(height, width)
    resized_width = max(1, int(np.floor(width * scale + 0.5)))
    resized_height = max(1, int(np.floor(height * scale + 0.5)))
    resampling = Image.Resampling.BOX if scale < 1 else Image.Resampling.NEAREST
    resized = Image.fromarray(tight).resize((resized_width, resized_height), resample=resampling)
    pixels = np.ceil(np.asarray(resized)).astype(np.uint8)

    x = (canvas - resized_width) // 2
    y = (canvas - resized_height) // 2
    result[y:y+resized_height, x:x+resized_width] = pixels
    return result
```

<div dir="rtl" align="right" style="direction: rtl; text-align: right;">
<p>ابتدا کوچک‌ترین مستطیل دور پیکسل‌های سفید را می‌گیریم. داخل بوم <bdi dir="ltr">32 × 32</bdi> و بعد از کنار گذاشتن حاشیه‌ی سه‌پیکسلی، فضایی به اندازه‌ی <bdi dir="ltr">26 × 26</bdi> داریم. برای عرض و ارتفاع از یک ضریب مشترک استفاده می‌کنیم تا شکل کشیده نشود. طول ضلع بزرگ‌تر به ۲۶ می‌رسد و ضلع دیگر به نزدیک‌ترین تعداد پیکسل گرد می‌شود. کوچک‌ترین اندازه یک پیکسل است.</p>

```python
# B2 — Define the normalization and geometry checks for a saved file
B2_OUTPUT = OUTPUT / "B2"
B2_OUTPUT.mkdir(exist_ok=True)
b2_geometry = []


def save_normalized(source_path, target_path, identity):
    with Image.open(source_path) as image:
        source = np.asarray(image.convert("L"))
    normalized = normalize_character(source, **B2_CONFIG)
    source_box = foreground_bounds(source)
    result_box = foreground_bounds(normalized)
    if source_box is None or result_box is None:
        raise ValueError(f"Expected a non-empty supplied crop or template: {source_path}")
    _, _, right, bottom = source_box
    width, height = right - source_box[0], bottom - source_box[1]
    x, y, right, bottom = result_box
    nw, nh = right - x, bottom - y
    size = B2_CONFIG["canvas"]
    pads = x, y, size - right, size - bottom
    scale = (size - 2 * B2_CONFIG["margin"]) / max(width, height)
    assert max(nw, nh) == size - 2 * B2_CONFIG["margin"]
    assert abs(nw - width * scale) < 1 and abs(nh - height * scale) < 1
    assert min(pads) >= B2_CONFIG["margin"]
    assert abs(pads[0] - pads[2]) <= 1 and abs(pads[1] - pads[3]) <= 1
    Image.fromarray(normalized).save(target_path)
    b2_geometry.append({
        **identity, "file": target_path.relative_to(OUTPUT).as_posix(),
        "tight_width": width, "tight_height": height,
        "content_width": nw, "content_height": nh, "scale": round(scale, 6),
        "margin_left": pads[0], "margin_top": pads[1],
        "margin_right": pads[2], "margin_bottom": pads[3],
    })
    return normalized
```

```python
# B3 — Compute normalized cross-correlation using NumPy
def ncc(image, template):
    """Score a 2D image against one 2D template or a 3D stack of templates."""
    image = np.asarray(image, dtype=np.float64)
    template = np.asarray(template, dtype=np.float64)
    if image.ndim != 2 or template.ndim not in (2, 3) or template.shape[-2:] != image.shape:
        raise ValueError("NCC requires an image and templates with the same spatial shape.")
    if not image.size:
        raise ValueError("NCC requires a non-empty spatial shape.")

    image_centered = image - image.mean()
    template_centered = template - template.mean(axis=(-2, -1), keepdims=True)
    numerator = np.sum(image_centered * template_centered, axis=(-2, -1))
    image_energy = np.sum(image_centered**2)
    template_energy = np.sum(template_centered**2, axis=(-2, -1))
    denominator = np.sqrt(image_energy * template_energy)
    scores = np.full_like(numerator, np.nan)
    np.divide(numerator, denominator, out=scores, where=denominator > 0)
    return np.clip(scores, -1, 1)
```

```python
# B3 — Choose the best template per class and handle ties
def recognize_character(mask, templates, config):
    """Rank character classes for an already normalized mask."""
    pixels = np.asarray(mask)
    size = config["canvas"]
    if pixels.shape != (size, size):
        raise ValueError(f"Expected a normalized {size}x{size} character mask.")
    scores = ncc(pixels, templates["pixels"])
    best_by_label = {}
    for row, score in zip(templates["rows"], scores):
        if not np.isfinite(score):
            continue
        label, template_id = row["label"], row["template_id"]
        previous = best_by_label.get(label)
        if (previous is None or score > previous["score"]
                or (score == previous["score"] and template_id < previous["template_id"])):
            best_by_label[label] = {
                "label": label, "score": float(score), "template_id": template_id,
            }
    ranking = sorted(best_by_label.values(), key=lambda result: (-result["score"], result["label"]))
    if not ranking:
        return None
    return {**ranking[0], "ranking": ranking}
```

```python
# B3 — Define crop recognition and its failure handling
BOX_FIELDS = ("sample_id", "character_index", "x", "y", "width", "height")
CHAR_FIELDS = (*BOX_FIELDS, "predicted_label", "match_score", "template_id")
STRING_FIELDS = ("sample_id", "split", "prediction", "status", "error",
                 "segment_count", "mean_match_score")


def classify_segments(boxes, load_crop, templates, config):
    if not boxes:
        return [], [], "No character boxes"
    results, rankings = [], []
    for box in boxes:
        i = int(box["character_index"])
        try:
            result = recognize_character(load_crop(i), templates, config)
        except (OSError, ValueError) as exc:
            message = (exc.strerror or type(exc).__name__) if isinstance(exc, OSError) else str(exc)
            return [], [], f"Crop {i}: {message}"
        if result is None:
            return [], [], f"Crop {i}: no valid NCC score (empty or constant)"
        results.append({**{key: box[key] for key in BOX_FIELDS},
                        "predicted_label": result["label"],
                        "match_score": result["score"],
                        "template_id": result["template_id"]})
        rankings.append(result["ranking"])
    return results, rankings, ""


def load_normalized_crop(root, index):
    with Image.open(root / f"{index:03d}.png") as image:
        return np.asarray(image.convert("L"))
```

```python
# B4 — Compute the edit-distance table
from collections import Counter


def edit_distance_table(reference, prediction):
    """Cost to transform every reference prefix into every prediction prefix."""
    n, m = len(reference), len(prediction)
    costs = np.zeros((n + 1, m + 1), dtype=np.int32)
    costs[:, 0] = np.arange(n + 1)
    costs[0, :] = np.arange(m + 1)
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            substitution = int(reference[i - 1] != prediction[j - 1])
            costs[i, j] = min(costs[i - 1, j - 1] + substitution,
                              costs[i - 1, j] + 1, costs[i, j - 1] + 1)
    return costs


def levenshtein(reference, prediction):
    return int(edit_distance_table(reference, prediction)[-1, -1])
```

```python
# B4 — Trace the alignment back to individual edits
def align_edits(reference, prediction):
    costs = edit_distance_table(reference, prediction)
    i, j = len(reference), len(prediction)
    alignment = []
    # Deterministic ties: diagonal, then deletion, then insertion.
    while i or j:
        changed = i > 0 and j > 0 and reference[i - 1] != prediction[j - 1]
        if i and j and costs[i, j] == costs[i - 1, j - 1] + int(changed):
            alignment.append((i - 1, j - 1, reference[i - 1], prediction[j - 1],
                              "substitution" if changed else "correct"))
            i -= 1
            j -= 1
        elif i and costs[i, j] == costs[i - 1, j] + 1:
            alignment.append((i - 1, None, reference[i - 1], "", "deletion"))
            i -= 1
        else:
            alignment.append((None, j - 1, "", prediction[j - 1], "insertion"))
            j -= 1
    return int(costs[-1, -1]), alignment[::-1]
```

```python
# B4 — Define the CSV columns and load reference labels
STRING_RESULT_FIELDS = (*STRING_FIELDS, "reference", "edit_distance",
                        "reference_length", "exact_correct")
ALIGN_FIELDS = ("sample_id", "reference_index", "prediction_index",
                "reference_char", "predicted_char", "operation")
METRIC_FIELDS = ("split", "n", "exact_accuracy", "cer", "seg_precision",
                 "seg_recall", "count_agreement", "failures",
                 "exact_strings", "edits", "reference_characters")


def read_b4_csv(path):
    with path.open(encoding="utf-8-sig", newline="") as stream:
        return list(csv.DictReader(stream))


def labels_from_csv(path, manifest):
    rows = read_b4_csv(path)
    labels = {row["sample_id"]: row["text"] for row in rows}
    expected = {row["sample_id"] for row in manifest}
    if len(rows) != len(expected) or labels.keys() != expected or any(not s for s in labels.values()):
        raise ValueError(f"Missing, empty, or duplicate reference labels: {path}")
    return labels
```

```python
# B4 — Align saved predictions and load reference boxes
def eval_saved_strings(path, manifest, labels):
    predictions = read_b4_csv(path)
    by_id = {p["sample_id"]: p for p in predictions}
    selected = {row["sample_id"] for row in manifest}
    if len(by_id) != len(predictions) or not selected <= by_id.keys():
        raise ValueError(f"Missing or duplicate predictions: {path}")
    rows, aligned = [], []
    for item in manifest:
        sample_id = item["sample_id"]
        row = by_id[sample_id]
        reference, prediction = labels[sample_id], row["prediction"]
        status_ok = row["status"] in {"ok", "failed"} and (row["status"] == "ok" or not prediction)
        if item["split"] != row["split"] or not reference or not status_ok:
            raise ValueError(f"Invalid evaluation row for {sample_id}")
        distance, alignment = align_edits(reference, prediction)
        rows.append({**row, "reference": reference, "edit_distance": distance,
                     "reference_length": len(reference), "exact_correct": int(reference == prediction)})
        aligned.extend(dict(zip(ALIGN_FIELDS, (sample_id, *step))) for step in alignment)
    return rows, aligned


def reference_boxes_for(folder, manifest):
    by_sample = {}
    for item in manifest:
        sample_id = item["sample_id"]
        with (folder / f"{sample_id}.json").open(encoding="utf-8-sig") as stream:
            boxes = [tuple(c["box"]) for c in json.load(stream)["characters"]]
        if not boxes or any(len(box) != 4 or min(box[2:]) <= 0 for box in boxes):
            raise ValueError(f"Missing or invalid reference boxes for {sample_id}")
        by_sample[sample_id] = boxes
    return by_sample
```

```python
# B4 — Calculate exact accuracy, aggregate CER, and box metrics
def metrics_for(string_rows, split, predicted_boxes=None, reference_boxes=None):
    rows = [r for r in string_rows if split == "all" or r["split"] == split]
    edits = sum(row["edit_distance"] for row in rows)
    length = sum(row["reference_length"] for row in rows)
    exact = sum(row["exact_correct"] for row in rows)
    if not rows or not length:
        raise ValueError(f"No evaluable data for {split}")
    precision = recall = count_agreement = ""
    if predicted_boxes is not None:
        match_count = pred_count = ref_count = same_count = 0
        for row in rows:
            sample_id = row["sample_id"]
            predicted, reference = predicted_boxes[sample_id], reference_boxes[sample_id]
            pred_count += len(predicted)
            ref_count += len(reference)
            same_count += len(predicted) == len(reference)
            match_count += len(match_character_boxes(predicted, reference))
        precision = match_count / pred_count if pred_count else 0.0
        recall = match_count / ref_count
        count_agreement = same_count / len(rows)
    return {"split": split, "n": len(rows), "exact_accuracy": exact / len(rows),
            "cer": edits / length, "seg_precision": precision, "seg_recall": recall,
            "count_agreement": count_agreement, "failures": sum(r["status"] != "ok" for r in rows),
            "exact_strings": exact, "edits": edits, "reference_characters": length}
```

```python
# B4 — Prepare baseline paths and reusable crop helpers
B4_OUTPUT = OUTPUT / "B4"
B4_OUTPUT.mkdir(exist_ok=True)
baseline_output = OUTPUT / "baseline"
dev_manifest = [row for row in samples if row["split"] == "dev"]


def read_gray_file(path):
    with Image.open(path) as image:
        return np.asarray(image.convert("L"))


def crop_box(pixels, box):
    x, y, w, h = box
    return pixels[y:y+h, x:x+w]


def make_box_row(sample_id, i, box):
    return dict(zip(BOX_FIELDS, (sample_id, i, *box)))


def write_crop_set(root, sample_id, crops):
    folder = root / sample_id
    folder.mkdir(parents=True, exist_ok=True)
    for stale in folder.glob("*.png"):
        if stale.stem.isdigit() and int(stale.stem) >= len(crops):
            stale.unlink()
    for i, crop in enumerate(crops):
        Image.fromarray(crop).save(folder / f"{i:03d}.png")
```

```python
# B4 — Predict from a full mask with the shared recognizer
def baseline_mask(row):
    gray = read_gray_file(DATA / row["input_path"])
    return make_text_mask(gray, A3_CONFIG)


def segment_and_predict(manifest, destination, load_mask, *, save_pipeline_files=False):
    destination.mkdir(parents=True, exist_ok=True)
    if save_pipeline_files:
        for folder in ("masks", "overlays"):
            (destination / folder).mkdir(exist_ok=True)
    strings, characters, box_rows = [], [], []
    for item in manifest:
        sample_id = item["sample_id"]
        boxes, raw_crops, normalized = [], [], []
        mask = None
        try:
            mask = load_mask(item)
            boxes, cleaned = segment_characters(mask, B1_CONFIG)
            raw_crops = [crop_box(cleaned, box) for box in boxes]
            normalized = [normalize_character(crop, **B2_CONFIG) for crop in raw_crops]
            records = [make_box_row(sample_id, i, box) for i, box in enumerate(boxes)]
            results, _, error = classify_segments(records, normalized.__getitem__, B3_TEMPLATES, B2_CONFIG)
        except (OSError, ValueError) as exc:
            results, records, error = [], [], str(exc)
        if save_pipeline_files and mask is not None:
            Image.fromarray(mask).save(destination / "masks" / f"{sample_id}.png")
            gray = read_gray_file(DATA / item["input_path"])
            save_box_overlay(gray, boxes, destination / "overlays" / f"{sample_id}.png")
            write_crop_set(destination / "raw_crops", sample_id, raw_crops)
            write_crop_set(destination / "normalized_crops", sample_id, normalized)
        box_rows.extend(records)
        characters.extend(results)
        strings.append(prediction_row(item, results, len(boxes), error))
    write_b3_csv(destination / "segmentation_boxes.csv", BOX_FIELDS, box_rows)
    write_b3_csv(destination / "character_results.csv", CHAR_FIELDS, characters)
    write_b3_csv(destination / "string_predictions.csv", STRING_FIELDS, strings)
```

```python
# B4 — Build string rows and define the oracle-crop control
def prediction_row(item, results, segment_count, error):
    return {
        "sample_id": item["sample_id"], "split": item["split"],
        "prediction": "".join(row["predicted_label"] for row in results),
        "status": "failed" if error else "ok", "error": error, "segment_count": segment_count,
        "mean_match_score": float(np.mean([row["match_score"] for row in results])) if results else "",
    }


def predict_oracle(manifest, destination, oracle_root):
    destination.mkdir(parents=True, exist_ok=True)
    strings, characters = [], []
    character_fields = ("sample_id", "character_index", "predicted_label", "match_score", "template_id")
    for item in manifest:
        sample_id = item["sample_id"]
        paths = sorted((oracle_root / sample_id).glob("*.png"))
        results = []
        error = ""
        try:
            if not paths or [path.stem for path in paths] != [f"{i:03d}" for i in range(len(paths))]:
                raise ValueError("Missing or non-consecutive oracle crops")
            for i, path in enumerate(paths):
                normalized = normalize_character(read_gray_file(path), **B2_CONFIG)
                best = recognize_character(normalized, B3_TEMPLATES, B2_CONFIG)
                if best is None:
                    raise ValueError(f"Oracle crop {i}: no valid NCC score")
                results.append({"sample_id": sample_id, "character_index": i,
                                "predicted_label": best["label"], "match_score": best["score"],
                                "template_id": best["template_id"]})
        except (OSError, ValueError) as exc:
            results, error = [], str(exc)
        characters.extend(results)
        strings.append(prediction_row(item, results, len(paths), error))
    write_b3_csv(destination / "character_results.csv", character_fields, characters)
    write_b3_csv(destination / "string_predictions.csv", STRING_FIELDS, strings)
```

```python
# B4 — Read the ground truth after prediction and define the scoring workflow
# Text ground truth first enters the final scoring path here.
reference_labels = labels_from_csv(reference_root / "labels.csv", samples)
diagnostic_labels = labels_from_csv(diagnostic_root / "evaluation" / "labels.csv", diagnostic_manifest)
assert all(reference_labels[sample_id] == label for sample_id, label in dev_labels.items())
main_reference_boxes = reference_boxes_for(reference_root / "annotations", samples)
diagnostic_reference_boxes = reference_boxes_for(diagnostic_root / "evaluation" / "annotations",
                                                   diagnostic_manifest)


def boxes_from_saved(folder, manifest):
    by_sample = {row["sample_id"]: [] for row in manifest}
    for row in read_b4_csv(folder / "segmentation_boxes.csv"):
        coords = tuple(int(row[key]) for key in ("x", "y", "width", "height"))
        box_list = by_sample[row["sample_id"]]
        if int(row["character_index"]) != len(box_list) or min(coords[2:]) <= 0:
            raise ValueError(f"Invalid saved box row: {row}")
        box_list.append(coords)
    return by_sample


def score_run(folder, manifest, labels, reference_boxes, *, has_predicted_boxes, splits):
    prediction_file = folder / "string_predictions.csv"
    if len(read_b4_csv(prediction_file)) != len(manifest):
        raise ValueError(f"Unbalanced prediction file: {prediction_file}")
    rows, alignments = eval_saved_strings(prediction_file, manifest, labels)
    boxes = boxes_from_saved(folder, manifest) if has_predicted_boxes else None
    scores = [metrics_for(rows, split, boxes, reference_boxes) for split in splits]
    write_b3_csv(folder / "string_results.csv", STRING_RESULT_FIELDS, rows)
    write_b3_csv(folder / "segmentation_results.csv", ALIGN_FIELDS, alignments)
    write_b3_csv(folder / "metrics.csv", METRIC_FIELDS, scores)
    return rows, alignments, scores, boxes
```

```python
# B1 — Save and show the box metric table
b1_metrics = []
for route in A2_ROUTES:
    for split in ("dev", "eval", "all"):
        rows = [row for row in b1_box_rows if row["route"] == route
                and (split == "all" or row["split"] == split)]
        matches = sum(row["matches"] for row in rows)
        predicted = sum(row["predicted"] for row in rows)
        reference = sum(row["reference"] for row in rows)
        count_equal = sum(row["count_equal"] for row in rows)
        b1_metrics.append({"route": route, "split": split, "n": len(rows),
                           "predicted": predicted, "reference": reference, "matches": matches,
                           "count_equal": count_equal,
                           "seg_precision": matches / predicted if predicted else 0.0,
                           "seg_recall": matches / reference,
                           "count_agreement": count_equal / len(rows)})

for name, rows in (("box_samples.csv", b1_box_rows), ("box_metrics.csv", b1_metrics)):
    with (B1_OUTPUT / name).open("w", encoding="utf-8", newline="") as stream:
        writer = csv.DictWriter(stream, fieldnames=list(rows[0]))
        writer.writeheader()
        writer.writerows(rows)

print("Route  Split  Matches  Predicted  Reference  Precision  Recall  Count agreement")
for r in b1_metrics:
    print(f"{r['route']:5}  {r['split']:5}  {r['matches']:7}  {r['predicted']:9}  "
          f"{r['reference']:9}  {r['seg_precision']:9.1%}  {r['seg_recall']:6.1%}  "
          f"{r['count_agreement']:15.1%}")
```

```python
# B5 — Gather the saved evidence for three distinct failures
B5_OUTPUT = OUTPUT / "B5"
B5_OUTPUT.mkdir(exist_ok=True)
failure_cases = (("A", "S011"), ("A", "S015"), (selected_route, "S001"))
failure_evidence = {}
failure_crop_rows = []
for route, sample_id in failure_cases:
    results, _, _, box_group = full_runs[route]
    result = next(row for row in results if row["sample_id"] == sample_id)
    assert not result["exact_correct"]
    crop_scores = [row for row in read_b4_csv(route_folders[route] / "character_results.csv")
                   if row["sample_id"] == sample_id]
    ref_boxes = main_reference_boxes[sample_id]
    matches = match_character_boxes(box_group[sample_id], ref_boxes)
    ref_by_prediction = {pi: (ri, iou) for pi, ri, iou in matches}
    for crop in crop_scores:
        pi = int(crop["character_index"])
        ri, iou = ref_by_prediction.get(pi, (None, None))
        failure_crop_rows.append({
            "route": route, **crop, "matched_reference_index": ri,
            "matched_reference_char": result["reference"][ri] if ri is not None else "",
            "matched_iou": iou,
        })
    failure_evidence[sample_id] = (route, result, box_group[sample_id], ref_boxes, crop_scores)

write_b3_csv(B5_OUTPUT / "failure_character_scores.csv", list(failure_crop_rows[0]), failure_crop_rows)
```

```python
# B5 — Define a four-panel inspection for one failed sample
def plot_failure(sample_id):
    route, result, boxes, ref_boxes, crop_scores = failure_evidence[sample_id]
    image = read_gray_file(DATA / f"inputs/{sample_id}.png")
    mask = read_gray_file(route_folders[route] / "masks" / f"{sample_id}.png")
    reference = read_gray_file(reference_root / "reference_masks" / f"{sample_id}.png")
    fig, axes = plt.subplots(2, 2, figsize=(8.55, 3.85), layout="constrained")
    fig.suptitle(f"{sample_id} | {route} | reference {result['reference']}  "
                 f"→  predicted {result['prediction']} ({result['edit_distance']} edits)", fontsize=12)
    for ax, pixels, panel_boxes, title in zip(
        axes.flat, (image, mask, mask, reference), ((), (), boxes, ref_boxes),
        ("Official input", "Generated mask", "Prediction boxes and NCC scores", "Reference mask and boxes"),
    ):
        show_evaluation_boxes(ax, pixels, panel_boxes)
        ax.set_title(title, fontsize=10)
    predicted_panel = axes[1, 0]
    predicted_panel.set_ylim(mask.shape[0] - 0.5, -5)
    for (x, y, w, h), score in zip(boxes, crop_scores, strict=True):
        predicted_panel.text(x + (w - 1) / 2, y - 0.8,
                             f"{score['predicted_label']} {float(score['match_score']):.2f}",
                             ha="center", va="bottom", fontsize=7, color="#165662",
                             bbox={"facecolor": "white", "edgecolor": "none", "pad": 0.7})
    fig.savefig(B5_OUTPUT / f"{sample_id}_{route}.png", dpi=220, facecolor="white")
    plt.show()
    plt.close(fig)
```

## فرمول‌ها و ساختار فایل خروجی

$$NCC(X,T)=\frac{\sum (X-\bar X)(T-\bar T)}{\sqrt{\sum(X-\bar X)^2\sum(T-\bar T)^2}}$$

$$Exact\ Accuracy=\frac{\#\{\hat y_i=y_i\}}{N}\qquad CER=\frac{\sum_i d(y_i,\hat y_i)}{\sum_i |y_i|}$$

⁦NCC⁩ در بازهٔ منفی یک تا یک قرار دارد و ⁦CER⁩ در صورت درج‌های زیاد می‌تواند از یک بیشتر شود
مقادیر را به صورت نسبت ذخیره کنید و برای نمایش درصد در ۱۰۰ ضرب کنید

| فایل | ستون‌های لازم |
|---|---|
| ⁦`segmentation_boxes.csv`⁩ | ⁦sample_id, character_index, x, y, width, height⁩ |
| ⁦`character_results.csv`⁩ | ستون‌های کادر همراه ⁦predicted_label, match_score, template_id⁩ |
| ⁦`string_predictions.csv`⁩ | ⁦sample_id, split, prediction, status, error, segment_count, mean_match_score⁩ |
| ⁦`string_results.csv`⁩ | ستون‌های پیش‌بینی همراه ⁦reference, edit_distance, reference_length, exact_correct⁩ |
| ⁦`segmentation_results.csv`⁩ | ⁦sample_id, reference_index, prediction_index, reference_char, predicted_char, operation⁩ |
| ⁦`metrics.csv`⁩ | ⁦split, n, exact_accuracy, cer, seg_precision, seg_recall, count_agreement, failures⁩ |
| ⁦`config.json`⁩ | روش‌ها و پارامترهای عمومی و بذرها |

در هم‌ترازی ویرایشی ⁦operation⁩ یکی از ⁦correct⁩ و ⁦substitution⁩ و ⁦insertion⁩ و ⁦deletion⁩ است
برای درج یا حذف اندیس متناظرِ ناموجود خالی می‌ماند
اندیس کاراکترها از صفر است و ⁦CSV⁩ها با ⁦UTF-8⁩ ذخیره شوند
اگر کادر یا متن مرجع یک شناسه موجود نباشد ارزیابی نامعتبر است و نباید با حذف شناسه ادامه دهید
